# Vertices list

```python
def extract_vertices(edges):
    vertices_list = []
    for u, v, distance in edges:
        if u not in vertices_list:
            vertices_list.append(u)
        if v not in vertices_list:
            vertices_list.append(v)
    return vertices_list
```

# Adjazenzliste

```python
def create_adjazenzliste(edges):
    adjazenzliste = {}
    for u,v,distance in edges:
        if u not in adjazenzliste: adjazenzliste[u] = []
        if v not in adjazenzliste: adjazenzliste[v] = []
        adjazenzliste[u].append((v,distance))
        adjazenzliste[v].append((u,distance))
    return adjazenzliste
```

#Adjazenzmatrix

```python
def create_adjazenzmatrix(edges):
    adjazenzmatrix = []
    #Name to index
    name_to_index = {}
    vertices = extract_vertices(edges)
    n = len(vertices)
    for i in range(n):
        node_name = vertices[i]
        name_to_index[node_name] = i
    #Matrix
    #Create matrix
    for i in range(n):
        row = []
        for j in range (n):
            row.append(float('inf'))
        adjazenzmatrix.append(row)
    for i in range(n):
        adjazenzmatrix[i][i] = 0
    for u,v,distance in edges:
        adjazenzmatrix[name_to_index[u]][name_to_index[v]] = distance
        adjazenzmatrix[name_to_index[v]][name_to_index[u]] = distance
    return adjazenzmatrix, name_to_index
```

# Prim's algorithm (Blatt 3): Find the shortest path to reach every node without creating a cycle

```python
def mst_prim(edges):
    matrix, name_to_index = create_adjazenzliste(edges)
    vertices = extract_vertices(edges)
    n = len(vertices)
    visited = [False] * n
    visited[0] = True
    mst_edges = []
    total_distance = 0
    for _ in range(n-1): #n-1 là số cạnh để liên kết n đỉnh mà ko chứa Kreis
        from_node_index = -1
        to_node_index = -1
        min_cost = float("inf")
        for i in range(n):
            if visited[i]:
                for j in range(n):
                    if not visited[j] and matrix[i][j] is not None:
                        if matrix[i][j] < min_cost: 
                            min_cost = matrix[i][j]
                            from_node_index = i
                            to_node_index = j
        # If the pair is chosen to be in MST: 
        if to_node_index != -1:
            visited[to_node_index] = True
            total_distance += min_cost
            mst_edges.append((vertices[from_node_index], vertices[to_node_index]))
    return mst_edges, total_distance
```
